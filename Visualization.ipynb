```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Student": ["Arun", "Bala", "Charan", "David", "Ezhil",
                "Farhan", "Gokul", "Hari"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 4],
    "Attendance": [60, 65, 75, 80, 85, 90, 95, 70],
    "Assignment_Marks": [40, 50, 60, 65, 75, 80, 95, 55],
    "Exam_Scores": [35, 45, 55, 65, 75, 85, 95, 50]
}

df = pd.DataFrame(data)

print("STUDENT DATASET")
print(df)

# Line Plot
plt.figure(figsize=(8, 5))
plt.plot(df["Student"], df["Exam_Scores"], marker="o")
plt.title("Student Exam Performance")
plt.xlabel("Students")
plt.ylabel("Exam Scores")
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(df["Student"], df["Exam_Scores"])
plt.title("Exam Scores by Student")
plt.xlabel("Students")
plt.ylabel("Exam Scores")
plt.show()

plt.figure(figsize=(8, 5))
plt.scatter(df["Study_Hours"], df["Exam_Scores"])
plt.title("Study Hours vs Exam Scores")
plt.xlabel("Study Hours")
plt.ylabel("Exam Scores")
plt.show()


plt.figure(figsize=(8, 5))
plt.hist(df["Exam_Scores"], bins=5, edgecolor="black")
plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Scores")
plt.ylabel("Frequency")
plt.show()



plt.figure(figsize=(8, 5))
sns.boxplot(y=df["Exam_Scores"])
plt.title("Exam Score Distribution")
plt.show()



plt.figure(figsize=(8, 5))
sns.histplot(df["Exam_Scores"], bins=5, kde=True)
plt.title("Exam Score Distribution with KDE")
plt.show()


df["Performance"] = pd.cut(
    df["Exam_Scores"],
    bins=[0, 50, 75, 100],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Performance")
plt.title("Student Performance Categories")
plt.show()

correlation = df[
    ["Study_Hours", "Attendance",
     "Assignment_Marks", "Exam_Scores"]
].corr()

plt.figure(figsize=(8, 5))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Feature Correlation Heatmap")
plt.show()
sns.pairplot(
    df,
    vars=["Study_Hours", "Attendance",
          "Assignment_Marks", "Exam_Scores"]
)
plt.show()

print("\nDATA VISUALIZATION COMPLETED")
```
