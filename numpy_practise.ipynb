```python


import numpy as np

# 1. Create a 2D NumPy array
# Columns: Study Hours, Attendance, Assignment Marks, Exam Scores

students = np.array([
    [2, 70, 65, 45],
    [3, 75, 70, 55],
    [4, 80, 75, 65],
    [5, 85, 80, 75],
    [6, 90, 90, 90],
    [7, 95, 95, 95]
])

print("========== STUDENT DATASET ==========")
print(students)

# 2. Shape, dimensions and size
print("\n========== ARRAY INFORMATION ==========")
print("Shape:", students.shape)
print("Dimensions:", students.ndim)
print("Total elements:", students.size)

# 3. Indexing
print("\n========== INDEXING ==========")
print("First student's study hours:", students[0, 0])
print("Third student's attendance:", students[2, 1])
print("Last student's exam score:", students[-1, 3])

# 4. Slicing
print("\n========== SLICING ==========")
print("First three students:\n", students[:3])
print("Study hours:", students[:, 0])
print("Exam scores:", students[:, 3])

# 5. Separate exam scores
exam_scores = students[:, 3]

# 6. Mathematical operations
print("\n========== MATHEMATICAL OPERATIONS ==========")
print("Original scores:", exam_scores)
print("Add 5:", exam_scores + 5)
print("Subtract 5:", exam_scores - 5)
print("Multiply by 2:", exam_scores * 2)
print("Divide by 2:", exam_scores / 2)

# 7. Statistical functions
print("\n========== STATISTICS ==========")
print("Mean:", np.mean(exam_scores))
print("Median:", np.median(exam_scores))
print("Maximum:", np.max(exam_scores))
print("Minimum:", np.min(exam_scores))
print("Total:", np.sum(exam_scores))
print("Standard Deviation:", np.std(exam_scores))

# 8. Boolean filtering
print("\n========== FILTERING ==========")

high_performers = exam_scores > 70
print("Boolean condition:", high_performers)
print("Scores above 70:", exam_scores[high_performers])

# Students with marks > 60 AND attendance >= 80
selected_students = students[
    (students[:, 3] > 60) &
    (students[:, 1] >= 80)
]

print("Students scoring > 60 with attendance >= 80:\n",
      selected_students)

# 9. Update exam scores
print("\n========== UPDATED SCORES ==========")
updated_scores = exam_scores + 5
print("Updated scores:", updated_scores)

# 10. Separate features X and target y
print("\n========== MACHINE LEARNING DATA PREPARATION ==========")

X = students[:, :3]
y = students[:, 3]

print("Features X:")
print(X)

print("\nTarget y:")
print(y)
```
