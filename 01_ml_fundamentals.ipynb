```python

import numpy as np
import matplotlib.pyplot as plt

# Training data
X = np.array([1, 2, 3, 4, 5, 6, 7, 8], dtype=float)
y = np.array([12, 18, 25, 31, 39, 45, 52, 59], dtype=float)

w = 0.0
b = 0.0

learning_rate = 0.001
epochs = 200
n = len(X)

loss_history = []

# Training loop
for epoch in range(epochs):

    # Forward pass: make predictions
    y_pred = w * X + b

    # Calculate mean squared error
    loss = np.mean((y_pred - y) ** 2)
    loss_history.append(loss)

    # Calculate gradients
    dw = (2 / n) * np.sum(X * (y_pred - y))
    db = (2 / n) * np.sum(y_pred - y)

    # Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("Learned weight:", round(w, 3))
print("Learned bias:", round(b, 3))
print("Final training loss:", round(loss_history[-1], 4))


hours = 7.5
predicted_score = w * hours + b

print("Predicted score:", round(predicted_score, 2))
```

```python

plt.figure(figsize=(8, 4))
plt.plot(loss_history)
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.title("Training Loss During Gradient Descent")
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 4))
plt.scatter(X, y, label="Actual scores")
plt.plot(X, w * X + b, label="Learned regression line")
plt.xlabel("Hours studied")
plt.ylabel("Exam score")
plt.legend()
plt.grid(True)
plt.show()
```
